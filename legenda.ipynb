```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model

data = {
    'Volume': [1000, 1500, 1500, 1500, 1600, 1600, 1600, 1600, 1600, 1800, 1800, 1800,
               2000, 2000, 2000, 2000, 2000, 1000, 1000, 1000, 1200, 1200, 1200, 1200,
               1200, 1400, 1400, 1400, 1500, 1500, 1500, 1500, 1500, 1600, 1800, 2000],
    'Weight': [790, 1160, 929, 865, 1140, 929, 1109, 1365, 1112, 1150, 980, 990,
               1112, 1252, 1326, 1330, 1365, 1280, 1119, 1328, 1584, 1428, 1365, 1415,
               1415, 1465, 1490, 1725, 1523, 1705, 1605, 1746, 1235, 1390, 1405, 1395],
    'CO2':    [99, 95, 95, 90, 105, 105, 90, 92, 98, 99, 99, 101,
               99, 94, 97, 97, 99, 104, 104, 105, 94, 99, 99, 99,
               99, 102, 104, 114, 109, 114, 115, 117, 104, 108, 109, 120]
}
df = pd.DataFrame(data)

model = linear_model.LinearRegression()
model.fit(df[['Volume', 'Weight']], df['CO2'])
print('coef :', model.coef_)
print('intercept :', model.intercept_)

volume_range = np.linspace(df['Volume'].min(), df['Volume'].max(), 10)
weight_range = np.linspace(df['Weight'].min(), df['Weight'].max(), 10)
volume_grid, weight_grid = np.meshgrid(volume_range, weight_range)
co2_pred = model.intercept_ + model.coef_[0] * volume_grid + model.coef_[1] * weight_grid

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(df['Volume'], df['Weight'], df['CO2'], color='red')
ax.plot_surface(volume_grid, weight_grid, co2_pred, alpha=0.5)
ax.set_xlabel('Volume')
ax.set_ylabel('Weight')
ax.set_zlabel('CO2')
plt.show()
```
